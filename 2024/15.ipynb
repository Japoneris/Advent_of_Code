# Part 1

```python
def init_puzzle():
    M = np.zeros((len(P1), len(P1[0])))
    S = None
    for idx, line in enumerate(P1):
        for jdx, c in enumerate(line):
            if c =="#":
                M[idx, jdx] = -1
            elif c == "O":
                M[idx, jdx] = 1
            elif c == "@":
                S = [idx, jdx]
    
    return M, S    



def print_puzzle(M):
    for line in M:
        for c in line:
            if c == -1:
                print("#", end="")
            elif c == 0:
                print(".", end="")
            else:
                print("o", end="")
        print()
    return
```

## Play

```python
M, S = init_puzzle()

dic_moves = {">": (0, 1), "<": (0, -1), "^": (-1, 0), "v": (1, 0)}

s0, s1 = S
for move in P2:
    dx, dy = dic_moves[move]
    # While loop until:
    #    - block (do nothing)
    #    - Empty (move all previous elements
    
    for i in range(1, 50):
        c = M[s0 + dx*i, s1 +dy * i]
        if c == -1:
            # No move
            break
            
        elif c == 0:
            # Move everything
            for j in range(i-1, 0, -1): # In reverse order
                M[s0+dx*(j+1), s1+dy*(j+1)] = M[s0+dx*j, s1+dy*j]
            
            M[s0, s1] = 0
            
            s0 = s0 + dx
            s1 = s1 + dy
            break
        # Else: Object, check next element
        
```

```python
def init_puzzle_v2():
    M = np.zeros((len(P1), 2*len(P1[0])))
    S = None
    for idx, line in enumerate(P1):
        for jdx, c in enumerate(line):
            if c =="#":
                M[idx, 2*jdx:2*(jdx+1)] = -1
                
            elif c == "O":
                M[idx, 2*jdx:2*(jdx+1)] = [1, 2]
                
            elif c == "@":
                S = [idx, 2*jdx]
    
    return M, S    


def print_puzzle_v2(M):
    for line in M:
        for c in line:
            if c == -1:
                print("#", end="")
            elif c == 0:
                print(".", end="")
            elif c == 1:
                print("[", end="")
            elif c == 2:
                print("]", end="")
            
        print()
    return
```

## Move the boxes ...

```python
def get_involved_elements(px, py, M, dx):
    lst = [py] # Pushing elements. The first one is not an object
    mem = [] # involved elements, ie elemnts
    for i in range(1, 100): # While loop equivalent
        lst_new = []
        
        pp = px + i * dx
        
        for p in lst:
            if M[pp, p] == 1:
                lst_new.extend([p, p+1])
                mem.append([pp, p]) # Add the two elements
                mem.append([pp, p+1])
                
                
            elif M[pp, p] == 2:
                lst_new.extend([p-1, p])
                mem.append([pp, p])
                mem.append([pp, p-1])
                
            elif M[pp, p] == -1:
                # STOP
                return [], False
                
        lst = sorted(set(lst_new))
        if len(lst) == 0: # No constrains above.
            return mem, True
            
```

```python
from collections import defaultdict


def part2(puzzle_input):
    grid, directions = puzzle_input.split('\n\n')
    grid = [list(row) for row in grid.split('\n')]
    m, n = len(grid), len(grid[0])
    for i in range(m):
        for j in reversed(range(n)):
            if grid[i][j] == '#':
                grid[i].insert(j, '#')
            if grid[i][j] == '.':
                grid[i].insert(j, '.')
            if grid[i][j] == '@':
                robot = (i, j*2)
                grid[i][j:j+1] = ['.', '.']
            if grid[i][j] == 'O':
                grid[i][j:j+1] = ['[', ']']

    for d in directions[:]:
        i, j = robot
        
        if d == '<':
            k = j-1
            while grid[i][k] == ']':
                k -= 2
            if grid[i][k] == '.':
                for l in range(k, j):
                    grid[i][l] = grid[i][l+1]
                robot = (i, j-1)

        elif d == '>':
            k = j+1
            while grid[i][k] == '[':
                k += 2
            if grid[i][k] == '.':
                for l in reversed(range(j+1, k+1)):
                    grid[i][l] = grid[i][l-1]
                robot = (i, j+1)

        elif d == '^':
            queue = {(i-1, j)}
            rows = defaultdict(set)
            while queue:
                x, y = queue.pop()
                match grid[x][y]:
                    case '#':
                        break
                    case ']':
                        rows[x] |= {y-1, y}
                        queue |= {(x-1, y), (x-1, y-1)}
                    case '[':
                        rows[x] |= {y, y+1}
                        queue |= {(x-1, y), (x-1, y+1)}
                    case '.':
                        rows[x].add(y)
            else:
                for x in sorted(rows):
                    for y in rows[x]:
                        grid[x][y] = grid[x+1][y] if y in rows[x+1] else '.'
                robot = (i-1, j)

        elif d== 'v':
            queue = {(i+1, j)}
            rows = defaultdict(set)
            while queue:
                x, y = queue.pop()
                match grid[x][y]:
                    case '#':
                        break
                    case ']':
                        rows[x] |= {y-1, y}
                        queue |= {(x+1, y), (x+1, y-1)}
                    case '[':
                        rows[x] |= {y, y+1}
                        queue |= {(x+1, y), (x+1, y+1)}
                    case '.':
                        rows[x].add(y)
            else:
                for x in sorted(rows, reverse=True):
                    for y in rows[x]:
                        grid[x][y] = grid[x-1][y] if y in rows[x-1] else '.'
                robot = (i+1, j)

    total = 0
    for i in range(m):
        for j in range(n*2):
            if grid[i][j] == '[':
                total += 100*i + j

    return total
```
