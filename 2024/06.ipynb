```python
with open("inputs/06", "r") as fp:
    data = fp.readlines()
    data = list(map(lambda x: x[:-1], data))
```

```python
data = """....#.....
.........#
..........
..#.......
.......#..
..........
.#..^.....
........#.
#.........
......#...""".split("\n")
```

#### Search the guard

```python
i0, j0 = 0, 0 
for idx, line in enumerate(data):
    j0 = line.find("^")
    if j0 != -1:
        i0 = idx
        break

print("Start position:", i0, j0)

i_mem, j_mem = i0, j0

dico_dir = {"^": (-1, 0), "v": (1, 0), ">": (0, 1), "<": (0, -1)}

turn_dir = {"^": ">", ">": "v", "v": "<", "<": "^"}

visit_map = [[False for _ in line] for line in data]
visit_map[i0][j0] = True
```

## Visit loop

```python
direction = "^"
cnt = 0
while True:
    cnt += 1
    try:
        dx, dy = dico_dir[direction]
        i1, j1 = i0 + dx, j0 + dy
        if (i1 == -1) | (j1 == -1):
            print("NEGATIVE")
            break
        
        if data[i1][j1] == "#":
            # Border just in front
            direction = turn_dir[direction]
            
        else:
            i0, j0 = i1, j1
            visit_map[i0][j0] = True
    
    except:
        # Go out of the loop
        print("OUTSIDE")
        break

print(sum(list(map(sum, visit_map))))
```

# Part 2

1. From part 1, extract all location where the guard will pass through
2. Test each position

```python
def is_loop(data):
    
    direction = "^"
    cnt = 0
    i0, j0 = i_mem, j_mem
    
    while True:
        cnt += 1
        if cnt > 130 * 130:
            # Assume this is a sufficient stopping condition
            return True 
            
        try:
            dx, dy = dico_dir[direction]
            i1, j1 = i0 + dx, j0 + dy
            if (i1 == -1) | (j1 == -1):
                return False
            
            if data[i1][j1] == "#":
                # Border just in front
                direction = turn_dir[direction]
                
            else:
                i0, j0 = i1, j1
        
        except:
            # Go out of the loop
            return False
```

```python
lst_trials = []
for idx, line in enumerate(visit_map):
    for jdx, c in enumerate(line):
        if c:
            if (idx != i_mem) | (jdx != j_mem):
                lst_trials.append((idx, jdx))
```

```python
data = list(map(lambda line: [x for x in line], data))
```

```python
S = 0
for (a, b) in lst_trials:
    data[a][b] = "#"
    if is_loop(data):
        S += 1

    data[a][b] = "."

print(S)
```
