# Corrupted inputs

```python
def get_valid(word):
    word = word[4:-1]
    res = word.split(",")
    
    if len(res) != 2:
        return 0
    
    a, b = res
    if a.isnumeric() & b.isnumeric():
        a, b = int(a), int(b)
        return a * b
        
    else:
        return 0
        
```

# Part 1

```python
with open("inputs/03", "r") as fp:
    data = fp.read()

S = 0
while True:
    i = data.find("mul(")
    if i == -1:
        # end
        break

    j = data[i:].find(")")
    if j == -1:
        #end
        break
        
    word = data[i:i+j+1]
    S += get_valid(word)
    data = data[i+4:]

print(S)
```

# Part2: Do and Don't

```python
with open("inputs/03", "r") as fp:
    data = fp.read()

lst = []

FLAG = True
dico = {True: "don't()", False: "do()"}

while True:
    i = data.find(dico[FLAG])
    if i == -1:
        if FLAG:
            # Append last segment if DO
            lst.append(data)
            
        break
    
    if FLAG:
        lst.append(data[:i])

    data = data[i:]
    FLAG = FLAG == False


data = "".join(lst)
```

```python
S = 0
while True:
    i = data.find("mul(")
    if i == -1:
        # end
        break

    j = data[i:].find(")")
    if j == -1:
        #end
        break
        
    word = data[i:i+j+1]
    S += get_valid(word)
    data = data[i+4:]

print(S)
```
