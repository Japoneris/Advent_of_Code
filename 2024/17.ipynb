```python
def process_input(data):
    """
    Process the input
    
    :rparam: (registers, program)
    """
    a, b, c, _, pro = data.split("\n")
    fx = lambda x: int(x.split(": ")[1])
    ra, rb, rc = fx(a), fx(b), fx(c)

    codes = list(map(int, pro.split(": ")[1].split(",")))

    return (ra, rb, rc), codes
    
    
```

```python
registers, program = process_input(data)
```

```python
def get_combo(operand, registers):
    ra, rb, rc = registers
    
    if operand <= 3:
        return operand
    elif operand == 4:
        return ra
    elif operand == 5:
        return rb
    elif operand == 6:
        return rc
    else:
        assert(False)
    
```

```python
def process_program(registers, code, verbose=False):
    
    pointer = 0
    ra, rb, rc = registers
    l = len(code)
    mem = []
    while pointer <= l-2:
        opcode, operand = code[pointer:pointer+2]
        if verbose:
            print("==== {} x {} ({})===".format(opcode, operand, pointer))
        if opcode == 0:
            # division of register A
            v = 2**get_combo(operand, (ra, rb, rc))
            ra = ra // v
            
        elif opcode == 1:
            rb = rb ^ operand
        
        elif opcode == 2:
            rb = get_combo(operand, (ra, rb, rc)) % 8
    
        elif opcode == 3:
            if ra != 0:
                # Move to literal value
                # TODO
                pointer = operand
                continue
                
        elif opcode == 4:
            rb = rb ^ rc
            
        elif opcode == 5:
            v = get_combo(operand, (ra, rb, rc)) % 8
            #print(v, end=",")
            mem.append(v)
            
        elif opcode == 6:
            v = 2**get_combo(operand, (ra, rb, rc))
            rb = ra // v
    
        elif opcode == 7:
            v = 2**get_combo(operand, (ra, rb, rc))
            rc = ra // v

        # Default case
        pointer += 2
    
    return (ra, rb, rc), mem
```

## Part 1 solution

```python
reg, mem = process_program(registers, program)
print("Part 1:", ",".join(list(map(str, mem))))
```
