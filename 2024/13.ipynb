# Process input

```python
with open("inputs/13" ,"r") as fp:
    data = fp.read()
```

```python
data = data.split("\n")
```

```python
blocks = []
mem = []
for line in data:
    if line == "":
        blocks.append(mem)
        mem = []
    else:
        mem.append(line)
```

```python
def process_button(line):
    line = line.split(": X+")[1]
    a, b = line.split(", Y+")
    return int(a), int(b)

def process_prize(line):
    line = line.split(": X=")[1]
    a, b = line.split(", Y=")
    return int(a), int(b)
    

def process_blocks(lst):
    A = process_button(lst[0])
    B = process_button(lst[1])
    P = process_prize(lst[2])
    return [A, B, P]
```

```python
blocks = list(map(process_blocks, blocks))
```

# Part 1

```python
def get_possible_values(a, b, p):
    """For X or Y, check possible i, j values such as:
    a*i + b*j = p
    """
    imax = p // a
    lst = []
    for i in range(0, imax+1):
        p1 = p - a*i
        if p1 % b == 0:
            lst.append((i, p1//b))
    
    return lst
        
    
```

```python
def get_cheapest_combination(block):
    lst0 = get_possible_values(block[0][0], block[1][0], block[2][0])
    lst1 = get_possible_values(block[0][1], block[1][1], block[2][1])

    lst0 = list(map(lambda x: "{}_{}".format(x[0], x[1]), lst0))
    lst1 = list(map(lambda x: "{}_{}".format(x[0], x[1]), lst1))
    lst = set(lst0) & set(lst1)

    lst_scores = []
    for elt in lst:
        a, b = elt.split("_")
        a, b = int(a), int(b)
        if (a > 100) | (b > 100):
            continue
            
        lst_scores.append((a*3+b, a, b))

    lst_scores = sorted(lst_scores, key=lambda x: x[0])
    if len(lst_scores) > 0:
        return lst_scores[0]
    else:
        return 0, 0, 0
```

```python
S = 0
for block in blocks:
    res = get_cheapest_combination(block)
    S += res[0]
```

# Part 2


```python
blocks_updated = []
dx = 10000000000000
for block in blocks:
    a, b = block[2]

    blocks_updated.append(block[:2] + [(a+dx, b+dx)])
```

```python
S = 0

for block in blocks_updated:
    (xa, ya), (xb, yb), (Tx, Ty) = block

    # Search valid CA for X
    mem = []
    for i in range(max(xa, xb)*10):
        left = (Tx - i *xa) % xb
        
        if left == 0:
            mem.append(i)
    
    if len(mem) == 0:
        continue
        
    a0, b0 = int(mem[0]), mem[1] - mem[0]
    # CA = a0 + i *b0
    
    # Search valid CA for Y
    mem = []
    for i in range(max(ya, yb)*10):
        left = (Ty - i * ya) % yb
        if left == 0:
            mem.append(i)

    if len(mem) == 0:
        continue
        
    a1, b1 = int(mem[0]), mem[1] - mem[0]
    # CA = a1 + i *b1

    
    # Join both information to get a CA that verify both
    mem0 = []
    mem1 = []
    
    for i in range(max(b0, b1)*4):
        mem0.append(a0 + i*b0)
        mem1.append(a1 + i*b1)
    
    common = sorted(set(mem0) & set(mem1))
    if len(common) == 0:
        continue
        
    # Loi pour les CA valides
    # CA = a2 + i *b2
    a2, b2 = int(common[0]), common[1] - common[0]



    # Search affine relationship for CB
    mem0 = []
    mem1 = []
    for i in range(10):
        ca = a2 + i * b2
        cb0 = (Tx - ca * xa) // xb
        cb1 = (Ty - ca * ya) // yb
        #print((Tx - ca * xa) % xb)
        #print((Ty - ca * ya) % yb)
        #print(cb0, cb1)
        mem0.append(cb0)
        mem1.append(cb1)
    
    a3, b3 = mem0[0], mem0[1] - mem0[0]
    a4, b4 = mem1[0], mem1[1] - mem1[0]
    
    left = (a4 - a3) % (b3 - b4)
    i_best = (a4 - a3) // (b3 - b4)
    print("Leftover ...", left)
    print(i_best)
    if left == 0:
        print("=======================")
        
        ca = a2 + i_best * b2
        cb = (Tx - ca * xa) // xb
        
        S += ca * 3 + cb
```
