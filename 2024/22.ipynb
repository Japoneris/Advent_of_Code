# Part 1: Just follow the instruction

In particular, each buyer's secret number evolves into the next secret number in the sequence via the following process:

-    Calculate the result of multiplying the secret number by 64. Then, **mix** this result into the secret number. Finally, **prune** the secret number.
-    Calculate the result of dividing the secret number by 32. **Round** the result down to the nearest integer. Then, **mix** this result into the secret number. Finally, **prune** the secret number.
-    Calculate the result of multiplying the secret number by 2048. Then, **mix** this result into the secret number. Finally, **prune** the secret number.

Each step of the above process involves mixing and pruning:

-    To **mix** a value into the secret number, calculate the bitwise XOR of the given value and the secret number. Then, the secret number becomes the result of that operation. (If the secret number is 42 and you were to mix 15 into the secret number, the secret number would become 37.)
-    To **prune** the secret number, calculate the value of the secret number modulo 16777216. Then, the secret number becomes the result of that operation. (If the secret number is 100000000 and you were to prune the secret number, the secret number would become 16113920.)


## Get FX

```python
def mix(secret, value):
    return secret ^ value

def prune(secret):
    return secret % 16777216

def mix_prune(secret, value):
    # operation after each step
    return prune(mix(secret, value))
    
    
```

```python
mix(42, 15), prune(100000000)
```

```python
def update_secret_number(secret):
    # Step 1
    secret = mix_prune(secret, secret * 64)
    secret = mix_prune(secret, secret // 32)
    return mix_prune(secret, secret * 2048)
    
```

## Our data

1. collect ALL sequences for each number
2. Identify all first buying price
3. Collect best sequence

#### Collect secret numbers

```python
LST = []

for example in data:
    secret = int(example[:-1])
    mem = [secret % 10] # Maybe remove this line if any issue
    for _ in range(2000):
        secret = update_secret_number(secret)
        mem.append(secret % 10)

    LST.append(mem)
```

#### Collect possible prices for each sequence

(takes some time to run)

```python
# Number of sequence to process
print(len(LST))
```

```python
LST_dico = []
S_seq = []
for idx, mem in enumerate(LST):
    print(idx, end="\r")
    dico = {}
    diff = np.diff(mem)

    # Loop in decreasing order to get "first buying price" without adding "if" condition
    for i in range(len(diff)-4, -1, -1):
        sq  = str(diff[i:i+4])
        dico[sq] = mem[i+4]
    
    S_seq.extend(list(dico))
    LST_dico.append(dico)
    
```

#### Find the best sequence

```python
S = set(S_seq)
print("Number of distinct sequences", len(S))

c_best = 0
for sq in S:
    cnt = 0
    for dico in LST_dico:
        if sq in dico:
            cnt += dico[sq]
    
    c_best = max(c_best, cnt)
    
```
