# Onsen towel

```python
with open("inputs/19", "r") as fp:
    data = fp.readlines()
    
```

```python
towels = data[0][:-1].split(", ")
```

```python
patterns = list(map(lambda x: x[:-1], data[2:]))
```

##### Organize towel by size

```python
dic_towels = {}
for t in towels:
    l = len(t)
    if l not in dic_towels:
        dic_towels[l] = [t]
    else:
        dic_towels[l].append(t)        

l_max = max(dic_towels)

print(l_max)
```

# Part 1: Check if one pattern is feasible

```python
def check_pattern(pattern):
    """Recursive function
    """
    if len(pattern) == 0:
        return True
        
    TF = False
    for li, vals in dic_towels.items():
        for v in vals:
            if pattern.startswith(v):
                res = check_pattern(pattern[li:])
                TF = TF | res
                if TF == True:
                    break
        if TF == True:
            break
            

    return TF
                
        
```

```python
cnt = 0
for p in patterns:
    if check_pattern(p):
        cnt += 1

print("Part 1:", cnt)
```

# Count the number of patterns

```python
def count_patterns(pattern):
    """
    Check subpatterns that can be created
    """
    l = len(pattern)
    arr = [0 for _ in range(l+1)]
    arr[0] = 1
    for i in range(l):
        for li, vals in dic_towels.items():
            for v in vals:
                if pattern[i:].startswith(v):
                    arr[i+li] = arr[i] + arr[i+li]
        

    return arr
                
        
```

```python
cnt = 0
for p in patterns:
    arr = count_patterns(p)
    cnt += arr[-1]

print("Part 2: ", cnt)
```
