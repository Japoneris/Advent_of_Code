| 7 | 8 | 9 |

```txt
    +---+---+
    | ^ | A |
+---+---+---+
| < | v | > |
+---+---+---+
```

````python
from itertools import permutations
# Useless, better to have same move near to each other to avoid extra cost
````

````python
dic_mapping_0 = {
    "7": (0, 0),
    "8": (0, 1),
    "9": (0, 2),
    "4": (1, 0),
    "5": (1, 1),
    "6": (1, 2),
    "1": (2, 0),
    "2": (2, 1),
    "3": (2, 2),
    "0": (3, 1),
    "A": (3, 2),
}

dic_mapping_1 = {
    "^": (0, 1),
    "A": (0, 2),
    "<": (1, 0),
    "v": (1, 1),
    ">": (1, 2),
}

dic_move = {
    "^": (-1, 0),
    "v": (1, 0),
    ">": (0, 1),
    "<": (0, -1)
}
````

````python
code = "805A"
````

````python
def is_valid_path_0(start, move_sequence):
    """Check if never encounter empty path

    Should bever reach (3, 0)
    
    :param start: (x, y)
    :param move_sequence: string like "^>>^"
    """
    x, y = start
    for m in move_sequence:
        dx, dy = dic_move[m]
        x += dx
        y += dy
        if (x == 3) & (y == 0):
            return False
    
    return True
    

def is_valid_path_1(start, move_sequence):
    """Check if never encounter empty path

    Should bever reach (0, 0)
    
    :param start: (x, y)
    :param move_sequence: string like "^>>^"
    """
    x, y = start
    for m in move_sequence:
        dx, dy = dic_move[m]
        x += dx
        y += dy
        if (x == 0) & (y == 0):
            return False
    
    return True
````

````python
def from_a_to_b_01(a, b, verbose=False):
    """
    Robot that control keypad to press last pad
    """
    x0, y0 = dic_mapping_0[a]
    x1, y1 = dic_mapping_0[b]
    
    # Delta that needs to be done
    dx, dy = x1 - x0, y1 - y0
    # Max steps possible
    dist = abs(dx) + abs(dy)

    # Generate all possible sequences
    S0, S1 = "", ""
    if dx > 0:
        S0 = "v" * dx
    else:
        S0 = "^" * abs(dx)

    if dy > 0:
        S1 = ">" * dy
    else:
        S1 = "<" * abs(dy)

    if verbose:
        print("Shortest path: {}".format(dist))
        print(S0 + S1)
    
    #lst = permutations(S0 + S1)
    #lst = list(map(lambda x: "".join(x), lst))
    #lst = list(set(lst)) # Get unique        
    lst = list(set([S0+S1, S1+S0]))
    
    if verbose:
        print("{} Permutations".format(len(lst)))
    
    lst_keep = []
    for path in lst:
        tf = is_valid_path_0((x0, y0), path)
        if tf:
            lst_keep.append(path)
    
    if verbose:
        print("==> {} Valid Permutations".format(len(lst_keep)))

    return lst_keep
````

````python
def from_a_to_b_11(a, b, verbose=False):
    """
    Robot that control keypad to press last pad
    """
    x0, y0 = dic_mapping_1[a]
    x1, y1 = dic_mapping_1[b]
    
    # Delta that needs to be done
    dx, dy = x1 - x0, y1 - y0
    # Max steps possible
    dist = abs(dx) + abs(dy)

    # Generate all possible sequences
    S0, S1 = "", ""
    if dx > 0:
        S0 = "v" * dx
    else:
        S0 = "^" * abs(dx)

    if dy > 0:
        S1 = ">" * dy
    else:
        S1 = "<" * abs(dy)

    if verbose:
        print("Shortest path: {}".format(dist))
        print(S0+S1)

    #lst = [S0+S1, S1+S0]
    lst = list(set([S0+S1, S1+S0]))
    
    #lst = permutations(S0 + S1)
    #lst = list(map(lambda x: "".join(x), lst))
    #lst = list(set(lst)) # Get unique        

    if verbose:
        print("{} Permutations".format(len(lst)))
    
    lst_keep = []
    for path in lst:
        tf = is_valid_path_1((x0, y0), path)
        if tf:
            lst_keep.append(path)
            
    if verbose:
        print("==> {} Valid Permutations".format(len(lst_keep)))

    return lst_keep
````

````python
def compute_path_01(code):
    """Get all path for the full code
    """
    lst_full = [""]
    for a, b in zip("A"+code[:-1], code):
        seq = from_a_to_b_01(a, b)
        lst_new = []
        for s0 in lst_full:
            for s1 in seq:
                lst_new.append(s0 + s1 + "A")

        lst_full = lst_new
    return lst_full
            
````

````python
def compute_path_11(code):
    """Get all path for the full code
    """
    lst_full = [""]
    for a, b in zip("A"+code[:-1], code):
        seq = from_a_to_b_11(a, b)
        lst_new = []
        for s0 in lst_full:
            for s1 in seq:
                lst_new.append(s0 + s1 + "A")

        lst_full = lst_new
        
    return lst_full
            
````

````python

numpad = {'7': (0, 0), '8': (1, 0), '9': (2, 0), '4': (0, 1), '5': (1, 1), '6': (2, 1), '1': (0, 2), '2': (1, 2), '3': (2, 2), '0': (1, 3), 'A': (2, 3)}
dirpad = {'^': (1, 0), 'A': (2, 0), '<': (0, 1), 'v': (1, 1), '>': (2, 1)}
keypad_dirs = {}
# Compute / Store best move to move from X to Y
for num1, loc1 in numpad.items():
    for num2, loc2 in numpad.items():
        num_pair = (num1, num2)
        if loc1[0] == loc2[0]: 
            keypad_dirs[num_pair] = ('v' if loc1[1] < loc2[1] else '^') * abs(loc1[1] - loc2[1])
        elif loc1[1] == loc2[1]: 
            keypad_dirs[num_pair] = ('>' if loc1[0] < loc2[0] else '<') * abs(loc1[0] - loc2[0])
        else:
            if loc1[0] == 0 and loc2[1] == 3: 
                keypad_dirs[num_pair] = '>' * (loc2[0] - loc1[0]) + 'v' * (loc2[1] - loc1[1])
            elif loc1[1] == 3 and loc2[0] == 0: 
                keypad_dirs[num_pair] = '^' * (loc1[1] - loc2[1]) + '<' * (loc1[0] - loc2[0])
            else: 
                keypad_dirs[num_pair] = [('v' if loc1[1] < loc2[1] else '^') * abs(loc1[1] - loc2[1]), ('>' if loc1[0] < loc2[0] else '<') * abs(loc1[0] - loc2[0])]

# Compute / Store best move to move from X to Y
# Avoid hole move
for dir1, loc1 in dirpad.items():
    for dir2, loc2 in dirpad.items():
        dir_pair = (dir1, dir2)
        if loc1[0] == loc2[0]: 
            keypad_dirs[dir_pair] = ('v' if loc1[1] < loc2[1] else '^') * abs(loc1[1] - loc2[1])
        elif loc1[1] == loc2[1]: 
            keypad_dirs[dir_pair] = ('>' if loc1[0] < loc2[0] else '<') * abs(loc1[0] - loc2[0])
        else:
            if loc1[0] == 0 and loc2[1] == 0: keypad_dirs[dir_pair] = '>' * (loc2[0] - loc1[0]) + '^'
            elif loc1[1] == 0 and loc2[0] == 0: keypad_dirs[dir_pair] = 'v' + (loc1[0] - loc2[0]) * '<'
            else: keypad_dirs[dir_pair] = [('v' if loc1[1] < loc2[1] else '^') * abs(loc1[1] - loc2[1]), ('>' if loc1[0] < loc2[0] else '<') * abs(loc1[0] - loc2[0])]

# Storage
known_sequences = {}

def next_robot(new_sequence, level):
    if (new_sequence, level) in known_sequences: 
        return known_sequences[(new_sequence, level)]
        
    if level == 26:
        #for part 1, just change '26' to '3'
        n = len(new_sequence)
    else:
        n = 0
        for i, c in enumerate(new_sequence):
            presses = keypad_dirs[('A' if i == 0 else new_sequence[i - 1], c)]
            n += min(next_robot(presses[0] + presses[1] + 'A', level + 1), next_robot(presses[1] + presses[0] + 'A', level + 1)) if isinstance(presses, list) else next_robot(presses + 'A', level + 1)
    
    known_sequences[(new_sequence, level)] = n
    return n

lst_codes = ["805A", "983A", "149A", "413A", "582A"]
print(sum([next_robot(code, 0) * int(code[:-1]) for code in lst_codes]))
````
