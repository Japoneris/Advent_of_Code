## Create bidirectionnal dictionnary

```python
data = """kh-tc
qp-kh
de-cg
ka-co
yn-aq
qp-ub
cg-tb
vc-aq
tb-ka
wh-tc
yn-cg
kh-ub
ta-co
de-co
tc-td
tb-wq
wh-td
ta-ka
td-qp
aq-cg
wq-ub
ub-vc
de-ta
wq-aq
wq-vc
wh-yn
ka-de
kh-ta
co-tc
wh-qp
tb-vc
td-yn""".split("\n")
data = list(map(lambda x: x + "\n", data))
```

```python
dico = {}
for line in data:
    a, b = line[:-1].split("-")
    
    if a not in dico:
        dico[a] = []
        
    if b not in dico:
        dico[b] = []

    dico[a].append(b)
    dico[b].append(a)
    

for key, val in dico.items():
    dico[key] = set(val)
```

# Part 1: get group of three users

```python
lst = []
for u0, nei in dico.items():
    for u1 in nei:
        SC = nei & dico[u1]
        for u2 in SC:
            if (u2 != u1) & (u2 != u0) & (u1 != u0):
                lst.append("-".join(sorted([u0, u1, u2])))


lst = set(lst)
```

```python
cnt = 0
for triplet in lst:
    a, b, c = triplet.split("-")
    if (a[0] == "t") | (b[0] == "t") | (c[0] == "t"):
        cnt += 1 
```

```python
print(len(lst))
print("Part 1 result:", cnt)
```

# Part 2: find max group

##### Maximum group size

```python
lmax = max(list(map(len, dico.values())))
print(lmax)
```

## Incremental

```python
# First, all elements are alone
# [elements, candidate set]
lst = [([c], val) for c, val in dico.items()]
print("START")
lst_stop = []

for _ in range(lmax -1):
    print(len(lst_stop), len(lst))
    lst_new = []
    for items, vals in lst:
        for vi in vals:
            lst_new.append((items + [vi], vals & dico[vi]))

    lst_stop.extend(list(filter(lambda x: len(x[1]) == 0, lst_new)))
    lst = list(filter(lambda x: len(x[1]) > 0, lst_new))

    dic_clean = {}
    for c, s in lst:
        dic_clean[str(sorted(c))] = (c, s)
    
    lst = list(dic_clean.values())
    
```

```python
dic_clean = {}
for c, s in lst_stop:
    dic_clean[str(sorted(c))] = c

lst = list(dic_clean.values())
```
