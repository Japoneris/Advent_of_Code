```python
with open("inputs/25", "r") as fp:
    data = fp.read()
```

```python
import numpy as np
```

```python
blocks = data[:-1].split("\n\n")
```

```python
def block_to_matrix(x):
    lines = x.split("\n")
    mat = np.zeros((7, 5))
    for idx, line in enumerate(lines):
        for jdx, c in enumerate(line):
            if c == "#":
                mat[idx, jdx] = 1
    return mat.astype(int)
```

```python
locks = list(filter(lambda x: x.startswith("#####"), blocks))
keys = list(filter(lambda x: x.startswith("#####") == False, blocks))


len(locks), len(keys)
```

```python
locks = list(map(block_to_matrix, locks))

keys = list(map(block_to_matrix, keys))
```

```python
def key_to_pin_height(block):
    return block.sum(axis=0) - 1    
```

```python
locks_code = list(map(key_to_pin_height, locks))
key_code = list(map(key_to_pin_height, keys))
```

```python
lst = []
for idx, lock in enumerate(locks_code):
    for jdx, key in enumerate(key_code):
        tot = lock + key
        #if (tot == 6).all():
        if (tot <= 5).all():
            print("{} match {}".format(idx, jdx))
            lst.append((idx, jdx))
```
