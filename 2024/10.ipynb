```python
import numpy as np
```

```python
with open("inputs/10", "r") as fp:
    data = fp.readlines()
```

```python
data = list(map(lambda x: x[:-1], data))
```

#### Convert map to numpy to simplify problem

```python
M = []
for line in data:
    M.append(list(map(int, line)))

M = np.array(M)
```

#### Add border to avoid border issues

```python
M.shape
```

```python
l = len(M)
T = -np.ones((l+2, l+2))
T[1:-1, 1:-1] = M

M = T
```

#### Identify starting position

```python
Start = np.array(np.where(M == 0)).T
```

# Part 1, explore all path

```python
def count_path(i0, j0, M, version="v1"):
    lst = [[i0, j0]]
    for v in range(1, 10):
        lst_new = []
        for dx, dy in [(0, 1), (0, -1), (-1, 0), (1, 0)]:
            for a, b in lst:
                if M[a+dx, b+dy] == v:
                    lst_new.append((a+dx, b+dy))
                    
        lst = lst_new

    # Remove duplicate ending locations
    if version == "v1":
        lst = set(list(map(lambda x: "{}_{}".format(x[0], x[1]), lst)))
    
    
    return len(lst)
```

```python
S = 0
for a, b in Start:
    S += count_path(a, b, M, version="v1")

print(S)
```

# Part 2

```python
S = 0
for a, b in Start:
    S += count_path(a, b, M, version="v2")

print(S)
```
