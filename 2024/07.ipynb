```python
with open("inputs/07", "r") as fp:
    data = fp.readlines()
```

```python
def process_line(line):
    a, vals = line[:-1].split(":")
    a = int(a)
    vals = list(map(int, vals.split()))
    return a, vals
    
```

```python
data = list(map(process_line, data))
```

# Part 1

```python
def recursive_evaluation(vals):
    mem = [vals[0]]
    for a in vals[1:]:
        mem = list(map(lambda x: x + a, mem)) + list(map(lambda x: x * a, mem))
        mem = list(set(mem))
    
    return mem
```

```python
S = 0
for a, vals in data:
    mem = recursive_evaluation(vals)
    if a in mem:
        S += a

print(S)
```

# Part 2

```python
def recursive_evaluation_p2(vals):
    mem = [vals[0]]
    for a in vals[1:]:
        mem = list(map(lambda x: x + a, mem)) + list(map(lambda x: x * a, mem)) + list(map(lambda x: int(str(x) + str(a)), mem))
        mem = list(set(mem))
    
    return mem
```

```python
S = 0
for a, vals in data:
    mem = recursive_evaluation_p2(vals)
    if a in mem:
        S += a

print(S)
```
