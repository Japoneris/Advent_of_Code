```python
import numpy as np
```

# Load + Process data

```python
with open("inputs/01", "r") as fp:
    data = fp.readlines()

def process_line(line):
    a, b = line[:-1].split()
    return int(a), int(b)
    

data = list(map(process_line, data))

data = np.array(data)
```

# Part 1

```python
V1 = np.array(sorted(data[:, 0]))
V2 = np.array(sorted(data[:, 1]))
S = np.abs(V1 - V2).sum()
print(S)
```

# Part 2

```python
dico = {}

for c in data[:, 1]:
    if c not in dico:
        dico[c] = 1
        
    else:
        dico[c] += 1
```

```python
S = 0
for v in V1:
    if v in dico:
        S += v * dico[v]

print(S)
```
