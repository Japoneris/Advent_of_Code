# Read data

```python
with open("inputs/05", "r") as fp:
    data = fp.readlines()
```

```python
rules = list(filter(lambda x: len(x) == 6, data))
orders = list(filter(lambda x: len(x) > 6, data))
```

```python
dic_rules = {}
for line in rules:
    a, b = line[:-1].split("|")
    a, b = int(a), int(b)
    if a not in dic_rules:
        dic_rules[a] = [b]
    else:
        dic_rules[a].append(b)
    
```

```python
def process_orders(line):
    vals = line[:-1].split(",")
    return list(map(int, vals))

orders = list(map(process_orders, orders))
```

# Part 1

```python
def check_an_order(order):
    """Reverse analysis:

    check if last element OK.
    """
    order = order[::-1]
    for idx, o in enumerate(order):
        if o not in dic_rules:
            # no constrains
            continue
        
        for a in dic_rules[o]:
            if a in order[idx+1:]:
                return False
    
    return True
        
        
```

```python
S = 0
lst_bad = [] # For P2
for order in orders:
    if check_an_order(order):
        l = len(order)
        S += order[l//2]
        
    else:
        lst_bad.append(order)

print(S)
        
    
```

# Part 2: Correct uncorrect ordering

```python
def check_rule(lst, rules):
    
    for a in rules:
        if a in lst:
            return False
    
    return True
    
```

```python
def correct_an_order(order):
    """
    All pages in the correct order are added first
    Pages that are not correct are inserted AFTER
    """
    mem = []
    out = []
    order = order[::-1]
    for idx, o in enumerate(order):
        if o not in dic_rules:
            # no constrains
            mem.append(o)
            continue

        FLAG = check_rule(order[idx+1:], dic_rules[o])
        if FLAG:
            mem.append(o)
        else:
            out.append(o)
    
    mem, out =  mem[::-1], out[::-1]
    #print(mem, out)
    if len(out) == 0:
        return mem
        
    # Now, try to insert the element rejected. At least at the end
    # Recursively, correct output first
    out = correct_an_order(out)
    for o in out[::-1]:
        for idx in range(len(mem)-1, -1, -1):
            if check_rule(mem[:idx], dic_rules[o]):
                mem = mem[:idx] + [o] + mem[idx:]
                break
    
    return mem
```

```python
S = 0
for order in lst_bad:
    re_order = correct_an_order(order)
    if check_an_order(re_order):
        l = len(re_order)
        S += re_order[l//2]
    else:
        print("ERRROR")
        print(order)

print(S)
```
