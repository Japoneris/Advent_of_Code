```python
with open("inputs/02", "r") as fp:
    data = fp.readlines()

def process_line(line):
    
    vals = line[:-1].split()
    return list(map(int, vals))
    
data = list(map(process_line, data))
```

# Part 1

Safe if:

- Increasing or decreasing
- Diff between 2 consecutive numbers [1, 2, 3]

Count the number of safe reports.

```python
def is_increasing(lst):
    """This function check that the diff is at least 1 and increasing
    """
    for a, b in zip(lst[:-1], lst[1:]):
        if a >= b:
            return False
    
    return True

def check_max_increment(lst):
    for a, b in zip(lst[:-1], lst[1:]):
        if b-a > 3:
            return False
    
    return True

def check_condition(lst):
    if is_increasing(lst):
        return check_max_increment(lst)
    elif is_increasing(lst[::-1]):
        return check_max_increment(lst[::-1])
    else:
        return False
    
```

```python
S = 0
for repport in data:
    if check_condition(repport):
        S += 1

print("Valid repports: {}".format(S))
```

# Part 2

Tolerate a single error (removal of one item)

```python
def create_sublist(lst):
    l = len(lst)
    sub = []
    for i in range(l):
        sub.append(lst[:i] + lst[i+1:])
    return sub
```

```python
S = 0
for repport in data:
    if check_condition(repport):
        S += 1
    else:
        sub = create_sublist(repport)
        for subrepport in sub:
            if check_condition(subrepport):
                S += 1
                break

print("Valid repports: {}".format(S))
```
