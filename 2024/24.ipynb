```python
def AND(x, y):
    return x & y

def OR(x, y):
    return x | y

def XOR(x, y):
    return x ^ y

dic_ops = {"OR": OR, "XOR": XOR, "AND": AND}
```

```python
while len(dic_P2) > 0:
    lst_del = []
    for key, (a, fx, b) in dic_P2.items():
        if (a in dic_P1) & (b in dic_P1):
            c = fx(dic_P1[a], dic_P1[b])
            dic_P1[key] = c
            lst_del.append(key)

    for key in lst_del:
        del(dic_P2[key])
```

## FX to evaluate quickly

```python
def create_puzzle():
    dic_P1 = {}
    for line in P1:
        a, b = line.split(": ")
        dic_P1[a] = int(b)
        
    dic_P2 = {}
    for line in P2:
        ops, v = line.split(" -> ")
        a, op, b = ops.split(" ")
        dic_P2[v] = (a, op, b)

    return dic_P1, dic_P2
```

```python
def evaluate(dic_P1, dic_P2):

    while len(dic_P2) > 0:
        lst_del = []
        for key, (a, op, b) in dic_P2.items():
            if (a in dic_P1) & (b in dic_P1):
                c = dic_ops[op](dic_P1[a], dic_P1[b])
                dic_P1[key] = c
                lst_del.append(key)
    
        for key in lst_del:
            del(dic_P2[key])

        if len(lst_del) == 0:
            # Stuck in a loop
            return False
            
    return True
```

```python
def switch(dic_P2, invert_list):
    for a, b in invert_list:
        dic_P2[a], dic_P2[b] = dic_P2[b], dic_P2[a]

    return
```

## For each key, retrieve dependencies

```python
def backtrack(key, dic_P2):
    lst = []
    lst_search = [key]
    while len(lst_search) > 0:
        a = lst_search.pop(0)
        lst.append(a)
        if a in dic_P2:
            t0, _, t1 = dic_P2[a]
            lst_search.append(t0)
            lst_search.append(t1)

    return sorted(set(lst))
    
```
