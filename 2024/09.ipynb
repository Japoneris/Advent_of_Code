```python
with open("inputs/09", "r") as fp:
    data = fp.read()

data = data[:-1]

data = list(map(int, data))
```

# Part 1

```python
lst_vals = []
lst_slot = []
for idx, v in enumerate(data):
    lst_slot.extend([idx % 2] * v)
    if idx % 2 == 0:
        lst_vals.extend([idx//2] * v)    
```

```python
l_tot = len(lst_vals)

for idx, v in enumerate(lst_slot[:l_tot]):
    if v == 0:
        lst_slot[idx] = lst_vals.pop(0)
    else:
        lst_slot[idx] = lst_vals.pop(-1)
        

lst_slot = lst_slot[:l_tot]
```

#### Hash

```python
S = 0
for idx, v in enumerate(lst_slot):
    S += idx * v

print(S)
```

# Part 2

Different strategy as contiguous moves.

Idea: mixed list

1. list of IDs
2. Number of free slot after it

```python
lst_empty = []
lst_full = []

cnt = 0

for idx, v in enumerate(data):
    if idx % 2 == 0:
        lst_full.append((cnt, [idx // 2] * v))
        
    else:
        lst_empty.append((cnt, v))
    
    cnt += v
```

```python
lst_new = []
for cnt, gp in lst_full[::-1]:
    l = len(gp)
    flag = True
    for idx, (ci, v) in enumerate(lst_empty):
        if ci > cnt:
            break
            
        if l <= v:
            lst_new.append((ci, gp))
            lst_empty[idx] = (ci + l, v-l) # Move ahead
            flag = False
            break
            
    if flag:
        lst_new.append((cnt, gp)) # No move
            
            
```

```python
S = 0
for idx, gp in lst_new:
    for i, v in enumerate(gp):
        S += (idx+i)*v

print(S)
```
