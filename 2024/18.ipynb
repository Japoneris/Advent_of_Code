# Load

```python
import numpy as np
```

```python
with open("inputs/18", "r") as fp:
    data = fp.readlines()
```

```python
def process_line(line):
    xx = line[:-1].split(",")
    return list(map(int, xx))
```

```python
def process_input(data):
    return list(map(process_line, data))
```

```python
coordinates = process_input(data)
```

```python
print(len(coordinates))
```

```python
def plot_grid(M):
    for line in M:
        for c in line:
            if c == 0:
                print(".", end="")
            else:
                print("#", end="")
        print()
```

# Part 1

```python
coords = coordinates[:1024]
size = 71
```

```python
def simulate_exit(coords, size):
    
    M = np.zeros((size, size))
    for a, b in coords:
        M[a, b] = 1

    
    S = np.ones((size, size)) * size**2
    
    # list of (position, score)
    lst = [((0, 0), 0)]
    while lst != []:
        pt, s = lst.pop(0)
        if (pt[0] < 0) | (pt[1] < 0) | (pt[0] >= size) | (pt[1] >= size):
            continue
    
        if M[pt[0], pt[1]] == 1:
            # Block
            continue
        
        if S[pt[0], pt[1]] > s:
            S[pt[0], pt[1]] = s
            for dx, dy in [(0, 1), (0, -1), (-1, 0), (1, 0)]:
                lst.append(((pt[0]+dx, pt[1]+dy), s+1))
        
        else:
            continue

    return S[-1, -1]
    
```

```python
s = simulate_exit(coords, size)

print("Part 1:", s)
```

Add other blocks, see if exit OK

```python
size = 71
maxi = size * size

for i in range(2500, len(coordinates)):
    coords = coordinates[:i]
    s = simulate_exit(coords, size)
    if s == maxi:
        print("Part 2: Step {}".format(i), coords[-1])
        break
```
