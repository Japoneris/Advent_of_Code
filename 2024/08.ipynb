```python
with open("inputs/08", "r") as fp:
    data = fp.readlines()
    data = list(map(lambda x: x[:-1], data))
```

# Part 1

1. Extract frequency location
2. For each frequency, calculate antinode location

#### Extract frequency

```python
dico = {}
for idx, line in enumerate(data):
    for jdx, c in enumerate(line):
        if c != ".":
            if c not in dico:
                dico[c] = []
            dico[c].append((idx, jdx))
```

### Antinode calculation

```python
lst = []
for vals in dico.values():
    for idx, (a0, b0) in enumerate(vals):
        for jdx, (a1, b1) in enumerate(vals):
            if idx == jdx:
                continue
            dx = a1 - a0
            dy = b1 - b0
            lst.append((a1+dx, b1+dy))    
            
```

### Filtering

1. Remove out of bounds
2. Remove duplicates

```python
l0, l1 = len(data), len(data[0])
lst = list(filter(lambda x: (x[0] >= 0) & (x[0] < l0) & (x[1] >= 0) & (x[1] < l1), lst))
```

### Answer

```python
print(len(set(list(map(lambda x: "{}_{}".format(x[0], x[1]), lst)))))
```

# Part 2

```python
lst = []

l0, l1 = len(data), len(data[0])

for vals in dico.values():
    for idx, (a0, b0) in enumerate(vals):
        for jdx, (a1, b1) in enumerate(vals):
            if idx == jdx:
                continue
            dx = a1 - a0
            dy = b1 - b0
            
            while True:
                a1, b1 = a1 + dx, b1 + dy
                if  (a1 >= 0) & (a1 < l0) & (b1 >= 0) & (b1 < l1):
                    lst.append((a1, b1))
                else:
                    break
                    
```

```python
for tmp in dico.values():
    lst.extend(tmp)
```

```python
print(len(set(list(map(lambda x: "{}_{}".format(x[0], x[1]), lst)))))
```
