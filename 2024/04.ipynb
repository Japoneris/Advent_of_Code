# Searching XMAS

```python
with open("inputs/04", "r") as fp:
    data = fp.readlines()

data = list(map(lambda x: x[:-1], data))
```

# Part 1

```python
def count_horizontal(line):
    c = 0
    while True:
        i = line.find("XMAS")
        if i == -1:
            break
        else:
            c += 1
            line = line[i+4:]
            
    return c

def count_horizontal_all(data):
    C = 0
    for line in data:
        C += count_horizontal(line[::-1])
        C += count_horizontal(line)
        
    return C

def count_vertical_all(data):
    # Transpose data to reuse horizontal function
    l1 = len(data[0])
    dataT = ["".join(list(map(lambda x: x[i], data))) for i in range(l1)]
    return count_horizontal_all(dataT)

def get_diag_word(data, i, j):
    return "".join([data[i+k][j+k] for k in range(4)])

def count_diagonal(data):
    l0, l1 = len(data), len(data[0])

    c = 0
    for i in range(l0-3):
        for j in range(l1-3):
            w = get_diag_word(data, i, j)
            if w == "XMAS":
                c += 1
    return c

def count_diagonal_all(data):
    c0 = count_diagonal(data)
    c1 = count_diagonal(data[::-1])
    c2 = count_diagonal(list(map(lambda x: x[::-1], data)))
    c3 = count_diagonal(list(map(lambda x: x[::-1], data[::-1])))

    return c0 + c1 + c2 + c3
```

```python
S = 0

S += count_horizontal_all(data)
S += count_vertical_all(data)
S += count_diagonal_all(data)
print(S)
```

# Part 2: Search for "MAS"

```python
def search_mas(data):
    l0, l1 = len(data), len(data[0])
    c = 0
    for i in range(l0-2):
        for j in range(l1-2):
            word = data[i][j] + data[i+2][j] + data[i+1][j+1] + data[i][j+2] + data[i+2][j+2]
            if word == "MMASS":
                c += 1
    return c
```

```python
def search_all_mas(data):
    c0 = search_mas(data)
    l1 = len(data[0])
    dataT = ["".join(list(map(lambda x: x[i], data))) for i in range(l1)]
    
    c1 = search_mas(dataT)
    c2 = search_mas(list(map(lambda x: x[::-1], data)))
    c3 = search_mas(list(map(lambda x: x[::-1], dataT)))
    return c0 + c1 + c2 + c3
```

```python
search_all_mas(data)
```
